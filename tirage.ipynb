```python
def reinitialiser():
    pioche = {"vert":[1, 6, 11, 16, 21, 26, 31, 36, 41, 46, 51, 56],
          "rose":[2, 7, 12, 17, 22, 27, 32, 37, 42, 47, 52, 57],
          "bleu":[3, 8, 13, 18, 23, 38, 33, 38, 43, 48, 53, 58],
          "rouge":[4, 9, 14, 19, 24, 29, 34, 39, 44, 49, 54, 59],
          "orange":[5, 10, 15, 20, 25, 30, 35, 40, 45, 50, 55, 60]}

    joueur1 = {"vert":[],
            "rose":[],
            "bleu":[],
            "rouge":[],
            "orange":[]}

    joueur2 = {"vert":[],
            "rose":[],
            "bleu":[],
            "rouge":[],
            "orange":[]}

    centre = {"vert":[],
            "rose":[],
            "bleu":[],
            "rouge":[],
            "orange":[]}
    
    defausse = {"vert":[],
            "rose":[],
            "bleu":[],
            "rouge":[],
            "orange":[]}
    return pioche, joueur1, joueur2, centre, defausse
```

```python
def deplacer_carte(entree, sortie, numero):
    for couleur, cartes in entree.items():
        if numero in cartes:
            cartes.remove(numero)
            sortie[couleur].append(numero)
            return
    print(f"La carte {numero} n'est pas dans ce dictionnaire !")
```

```python
import os

def clear():
    os.system('cls' if os.name == 'nt' else 'clear')

from IPython.display import clear_output
```

```python
import random

def tirer_carte(entree, sortie, couleur):
    if not entree[couleur]:
        print(f"Plus de cartes {couleur} dans la pioche !")
        return
    
    carte = random.choice(entree[couleur])
    entree[couleur].remove(carte)
    sortie[couleur].append(carte)
    return carte
```

```python
def initialisation_tirage(entre, sortie):
    tirer_carte(entre, sortie, "vert")
    tirer_carte(entre, sortie, "rose")
    tirer_carte(entre, sortie, "bleu")
    tirer_carte(entre, sortie, "rouge")
    tirer_carte(entre, sortie, "orange")
```

```python
def effectuer_tour(pioche, centre, defausse):
    from IPython.display import clear_output

    # Afficher le nombre de pions restants par couleur
    print("=== Pioche ===", flush=True)
    for couleur, cartes in pioche.items():
        print(f"{couleur} : {len(cartes)} pions restants", flush=True)
    # Afficher le centre
    print("\n=== Centre ===", flush=True)
    for couleur_c, cartes in centre.items():
        if cartes:
            print(f"{couleur_c} : {cartes}", flush=True)

    # Demander une couleur valide
    while True:
        couleur = input("\nChoisissez une couleur : ").strip().lower()
        if couleur == "end_game":
            return True
        if couleur in pioche and len(pioche[couleur]) > 0:
            break
        print("Couleur invalide ou pioche vide pour cette couleur !", flush=True)

    clear_output(wait=False)

    # Tirer une carte de la pioche vers le centre
    carte = tirer_carte(pioche, centre, couleur)
    if carte:
        print(f"Carte tirée : {carte} ({couleur})", flush=True)

    # Afficher le centre
    print("\n=== Centre ===", flush=True)
    for couleur_c, cartes in centre.items():
        if cartes:
            print(f"{couleur_c} : {cartes}", flush=True)

    # Demander un numéro valide
    while True:
        saisie = input("\nDonnez un numéro à déplacer du centre à la défausse : ").strip()
        if saisie == "end_game":
            return True
        try:
            numero = int(saisie)
            if any(numero in cartes for cartes in centre.values()):
                deplacer_carte(centre, defausse, numero)
                break
            print("Ce numéro n'est pas dans le centre !", flush=True)
        except ValueError:
            print("Numéro invalide !", flush=True)

    clear_output(wait=False)
    return False
```

```python
def boucle_jeu():

    from IPython.display import clear_output

    # Saisie des noms
    nom1 = input("Nom du Joueur 1 : ").strip()
    clear_output(wait=False)
    nom2 = input("Nom du Joueur 2 : ").strip()
    clear_output(wait=False)

    pioche, joueur1, joueur2, centre, defausse = reinitialiser()

    # Initialisation Joueur 1
    input(f"\n{nom1}, appuyez sur Entrée pour commencer votre tirage...")
    clear_output(wait=False)
    initialisation_tirage(pioche, joueur1)
    # Afficher le centre
    print(f"\n=== Pioche {nom1} ===", flush=True)
    for couleur_c, cartes in joueur1.items():
        if cartes:
            print(f"{couleur_c} : {cartes}", flush=True)
    input(f"\n{nom1}, appuyez sur Entrée quand vous avez terminé...")
    clear_output(wait=False)

    # Initialisation Joueur 2
    input(f"{nom2}, appuyez sur Entrée pour commencer votre tirage...")
    clear_output(wait=False)
    initialisation_tirage(pioche, joueur2)
    # Afficher le centre
    print(f"\n=== Pioche {nom2} ===", flush=True)
    for couleur_c, cartes in joueur2.items():
        if cartes:
            print(f"{couleur_c} : {cartes}", flush=True)
    input(f"\n{nom2}, appuyez sur Entrée quand vous avez terminé...")
    clear_output(wait=False)

    initialisation_tirage(pioche, centre)

    # Boucle de jeu
    joueurs = [(nom1, joueur1), (nom2, joueur2)]
    tour = 0

    while True:
        nom, joueur = joueurs[tour % 2]
        input(f"{nom}, c'est votre tour. Appuyez sur Entrée pour jouer...")
        clear_output(wait=False)

        fin = effectuer_tour(pioche, centre, defausse)  # on récupère si end_game a été tapé
        if fin:
            print("Fin de la partie !")
            break

        tour += 1
```

```python
boucle_jeu()
```
